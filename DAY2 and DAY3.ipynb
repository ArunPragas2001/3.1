```python
arr_a=np.full((50,50),50)  #dark gray square 
arr_b=np.full((50,50),200)  #light gray square 
#create a window with width of  8inches height of 4 inches 
plt.figure(figsize=(8,4))

plt.subplot(1,2,1) #1row , 2 columns,position 1(Left Side)
plt.imshow(arr_a, cmap='gray', vmin=0, vmax=255)
plt.title("Dark Square")

plt.subplot(1,2,2) #1row , 2 columns,position 1(Right Side)
plt.imshow(arr_b,cmap='gray',vmin=0,vmax=255)
plt.title("Light Square")
plt.show()
```
