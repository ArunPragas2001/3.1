```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img=cv2.imread("image1.png",cv2.IMREAD_GRAYSCALE).astype(np.float32)
```

```python
plt.figure(figsize=(6,5))
plt.imshow(img,cmap="gray", vmin=0, vmax=255)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
#-------------------------------------------------------------------------------------------------
#2.Brightness adjustment: add or substract a constanct
#-------------------------------------------------------------------------------------------------
brighter=img+60
brighter=np.clip(brighter,0,255)
brighter=brighter.astype(np.uint8)
```

```python
darker=img-60
darker=np.clip(darker,0,255)
darker=darker.astype(np.uint8)
```

```python
plt.figure(figsize=(15,4))
plt.subplot(1, 3, 1)
plt.imshow(darker, cmap="gray", vmin=0, vmax=255)
plt.title("darker (-60)")
plt.axis("off")


plt.subplot(1, 3, 2)
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("Original image ")
plt.axis("off")



plt.subplot(1, 3, 3)
plt.imshow(brighter, cmap="gray", vmin=0, vmax=255)
plt.title("Brighter(+60)")
plt.axis("off")
```

```python
min_Val = img.min()
max_Val = img.max()

print("Before stretching -> min:", min_Val, "max:", max_Val)

stretched = ((img - min_Val) / (max_Val - min_Val) * 255)

print("After stretching -> min:", stretched.min(), "max:", stretched.max())
```

```python
# Original image
plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("Original Image")
plt.axis("off")

# Stretched image
plt.subplot(1, 2, 2)
plt.imshow(stretched, cmap="gray", vmin=0, vmax=255)
plt.title("After Contrast Stretching")
plt.axis("off")

plt.show()
```

```python
#------------------------------------------------------------------
#4.Image negative/inversion 
#------------------------------------------------------------------
negative=255-img.astype(np.uint8)
img_uint8 = img.astype(np.uint8)
```

```python
#openCv's built-in bitwise_not gives the same result for 8-bit image
negative_cv2=cv2.bitwise_not(img_uint8)
```

```python


# Convert image to 8-bit
img_uint8 = img.astype(np.uint8)

# OpenCV negative
negative_cv2 = cv2.bitwise_not(img_uint8)

plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.imshow(img_uint8, cmap="gray", vmin=0, vmax=255)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(negative_cv2, cmap="gray", vmin=0, vmax=255)
plt.title("Negative (OpenCV)")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(255 - img_uint8, cmap="gray", vmin=0, vmax=255)
plt.title("Negative (Formula)")
plt.axis("off")

plt.show()
```

# 5. Threshold Based binarization using openCV's cv2.threshold 

```python
# Try threshold = 80
_, bin_img_80 = cv2.threshold(img_uint8, 80, 255, cv2.THRESH_BINARY)
```

```python
# Try threshold = 128
_, bin_img_100 = cv2.threshold(img_uint8, 100, 255, cv2.THRESH_BINARY)
```

```python
# Try threshold = 180
_, bin_img_180 = cv2.threshold(img_uint8, 180, 255, cv2.THRESH_BINARY)
```
