```python
import numpy as np
import matplotlib.pyplot as plt 
from PIL import Image 
```

```python
image_path="Image.jpg"
img=np.array(Image.open(image_path))
```

```python
plt.figure(figsize=(6,5))
if img.ndim == 2:
    plt.imshow(img,cmap="gray")
else:
    plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
height=img.shape[0]
width=img.shape[1]
dimension=img.shape[2]
print("Height:",height)
print("Width:",width)
print("Dimension:",dimension)
```

```python
if img.ndim==3:
    print("Color Image")
else:
    print("Grayscale Image")
```

```python
red_channel=img[:,:,0]
green_channel=img[:,:,1]
blue_channel=img[:,:,2]

```

```python
plt.figure(figsize=(5,4))
plt.imshow(red_channel,cmap="gray")
plt.title("Red channel")
plt.axis("off")
plt.show()
```

```python
row=50
col=50
print("Pixel before change:",img[row,col])
```

```python
modified_img=img.copy()
modified_img[row,col]=[255,0,0]
print("Pixel after change:",modified_img[row,col])
```

```python
roi_img=img.copy()
roi_img[100:200,50:250]=0
```

```python
plt.figure(figsize=(5,5))
plt.imshow(roi_img)
plt.title("Image with Black Squre ROI")

plt.show()
```

```python
cropped=img[100:250,100:250]
```

```python
plt.figure(figsize=(5,5))
plt.imshow(cropped)
plt.title("Cropped region")

plt.show()
```
