```python
import numpy as np
import matplotlib.pyplot as plt 
import cv2
from PIL import Image
```

```python
img_pil=Image.open("Scarlet-macaw-ara-macao.webp")
img_pil_array=np.array(img_pil)
```

```python
print(img_pil_array.shape)
```

```python
plt.imshow(img_pil_array)
plt.axis("off")
plt.show()
```

```python
print(img_pil_array.dtype)
```

```python
img_cv=cv2.imread("Scarlet-macaw-ara-macao.webp")
```

```python
print(img_cv.shape)
print(img_cv.dtype)
```

```python
img_cv_rgb=cv2.cvtColor(img_cv,cv2.COLOR_BGR2RGB)
plt.imshow(img_cv)
plt.show()
```

```python
img_gray=cv2.imread("Scarlet-macaw-ara-macao.webp",cv2.IMREAD_GRAYSCALE)
print(img_gray.shape)
```

```python
plt.imshow(img_gray,cmap="gray")
plt.axis("off")
plt.show()
```
